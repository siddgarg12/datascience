## Mixing strings with computations or data: f-Strings
f stands for formatted. 

```python
print(f"The temperature 75F in degrees celsius is {(75 - 32) * 5 / 9}C")

# formatting the numbers can be done using ":"
print(f"Isabel is {28/7:.1f} dog years old.") # with 1 decimal place
print(f"Isabel is {28/7:.2f} dog years old.") # with 2 decimal places
print(f"Isabel is {28/7:010} dog years old.") # with 10 signifincant digits including post decimal
print(f"Isabel is {28/7:,.1f} dog years old.") # floating point notation with commas
print(f"Isabel is {28/7:15,.1f} dog years old.") # floating point notation with commas, 15 spaces, default right aligned
print(f"Isabel is {28/7:<15,.1f} dog years old.") # floating point notation with commas, 15 spaces, left aligned
print(f"Isabel is {28/7:^15,.1f} dog years old.") # floating point notation with commas, 15 spaces, center aligned
print(f"Isabel is {28/7:*^15,.1f} dog years old.") # floating point notation with commas, 15 spaces, center aligned, filled with *
print(f"Isabel is {28/7:.1%} dog years old.") # percentage notation
print(f"Isabel is {28/7:.1e} dog years old.") # scientific notation

score = 7
print(f"Score: {score:03}")
```

# Visutalize data using Matplot.lib

```python
import matplotlib.pyplot as plt #matplotlib is a library for data visualization
sunny_days = [8, 10, 7, 14, 20, 18, 25, 19, 18, 14, 12, 7]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

avg_sunny_days = sum(sunny_days) / len(sunny_days)

plt.plot(months, sunny_days, marker="o")
plt.title("Sunny days")
plt.xlabel("Months", fontsize=12)
plt.ylabel("Sunny days", fontsize=12)
plt.axhline(y=avg_sunny_days, linestyle="--", color="orange", label="Average")
plt.legend()

plt.show()
```
